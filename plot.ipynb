```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Size of graph varying from 1 * 10 ^ 5 to 6 * 10 ^ 5 in the increments of 10 ^ 5 # of threads: 16
```

```python
x = np.array([
    100000, 200000, 300000, 400000, 500000, 600000
])

seq = np.array([
    (142184 + 139715 + 145609 + 140246 + 142425)/5,
    (305234 + 306693 + 310472 + 303140 + 302969)/5,
    (542565 + 543549 + 543841 + 542839 + 544244)/5,
    (772216 + 789326 + 791936 + 791443 + 790700)/5,
    (1016499 + 1020125 + 1014205 + 1015808 + 1019688)/5,
    (1256305 + 1254047 + 1254089 + 1251639 + 1254748)/5,
])

wf = np.array([
    (32211 + 31080 + 32605 + 33279 + 30932)/5,
    (49839 + 49195 + 47411 + 45376 + 46403)/5,
    (65958 + 67102 + 64856 + 67928 + 65657)/5,
    (86415 + 87634 + 84630 + 83467 + 87240)/5,
    (101266 + 103915 + 98551 + 102101 + 105323)/5,
    (127639 + 127109 + 120782 + 124499 + 125490)/5,
])

bar = np.array([
    (92399 + 92853 + 92357 + 92262 + 92334)/5,
    (174354 + 173110 + 173390 + 173321 + 175128)/5,
    (262098 + 260996 + 263465 + 261801 + 262798)/5,
    (348004 + 346208 + 347199 + 347505 + 347522)/5,
    (429909 + 431728 + 431058 + 429908 + 430238)/5,
    (516555 + 516244 + 520247 + 540175 + 518456)/5,
])

speed_up_wf = seq/wf
speed_up_bar = seq/bar

plt.figure()
plt.plot(x, seq, label = "Sequential")
plt.plot(x, bar, label = "Barrier")
plt.plot(x, wf, label = "Wait-Free")
plt.xlabel("No. of nodes")
plt.ylabel("Time (in µs)")
plt.legend()
plt.show()
```
