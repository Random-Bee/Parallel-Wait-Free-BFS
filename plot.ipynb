```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Size of graph varying from 1 * 10 ^ 5 to 6 * 10 ^ 5 in the increments of 10 ^ 5 # of threads: 16
```

```python
x = np.array([
    100000, 200000, 300000, 400000, 500000, 600000
])

seq = np.array([
    (138986 + 141636 + 139026 + 140915 + 142697)/5,
    (318584 + 324555 + 318702 + 320207 + 325564)/5,
    (553267 + 548667 + 544503 + 541110 + 543551)/5,
    (788396 + 789077 + 787938 + 788132 + 796122)/5,
    (1037590 + 1036038 + 1037083 + 1037259 + 1038563)/5,
    (1278252 + 1278961 + 1277425 + 1281363 + 1275457)/5,
])

wf = np.array([
    (31810 + 32244 + 31453 + 32102 + 33016)/5,
    (45275 + 48662 + 47636 + 47463 + 48931)/5,
    (67412 + 67174 + 66053 + 65497 + 67342)/5,
    (86296 + 87722 + 87724 + 79196 + 89285)/5,
    (107137 + 103662 + 99831 + 106582 + 105489)/5,
    (126858 + 124029 + 126587 + 125762 + 126548)/5,
])

speed_up = seq/wf

plt.figure()
plt.plot(x, seq, label = "Sequential")
plt.plot(x, wf, label = "Wait-Free")
plt.xlabel("No. of nodes")
plt.ylabel("Time (in µs)")
plt.legend()
plt.show()
```

```python
print("wf : ", wf)
print("seq : ", seq)
print("speed-up : ", speed_up)
```
